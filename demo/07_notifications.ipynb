# Notifications

Custom notifiers make notifications testable without desktop or SMTP credentials.

```python
from autocron.interface.notifications import NotificationManager

class RecordingNotifier:
    def __init__(self):
        self.messages = []
    def send(self, title, message, **kwargs):
        self.messages.append((title, message))
        return True

manager = NotificationManager()
notifier = RecordingNotifier()
manager.add_notifier("demo", notifier)
print(manager.notify("AutoCron", "completed", channels=["demo"]))
print(len(notifier.messages))
```
