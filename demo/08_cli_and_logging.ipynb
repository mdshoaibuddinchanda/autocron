# CLI and logging

The CLI exposes the durable task store as scriptable commands.

```python
import contextlib
import io
import json
import tempfile
from pathlib import Path
from autocron.interface.cli import main

with tempfile.TemporaryDirectory() as directory:
    root = Path(directory)
    script = root / "job.py"
    script.write_text("print('cli')", encoding="utf-8")
    database = root / "autocron.db"
    main(["--database", str(database), "schedule", str(script), "--every", "1h"])
    captured = io.StringIO()
    with contextlib.redirect_stdout(captured):
        main(["--database", str(database), "list", "--json"])
    print(json.loads(captured.getvalue())[0]["name"])
```
